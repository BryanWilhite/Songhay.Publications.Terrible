# responsive-image preparation

## Index splash (Q3 2026)


```python
import os

from IPython.display import HTML
from wand.color import Color
from wand.drawing import Drawing
from wand.image import Image

w1, h1 = 1600, 1200

rootDir = 'index-splash-Q3-2026'
```

## shared functions

```python
def display_img_html(img:Image, src:str, caption:str):
    label_style = f'display: block; margin: 0; padding: 0; position:absolute; top: 50px; left: 50px'
    img_style = 'display: block; margin: 0; padding: 0'

    kb = get_image_size_kb(src)

    html = f'''
    <div>
        <img src="{src}" style="{img_style}" />
        <span style="{label_style}">
            <code>{caption}: {img.width} x {img.height}</code>
            <br>
            <code>{kb}KB</code>
        </span>
    </div>
    '''
    display(HTML(html))

def get_image_size_kb(file_path: str):
    """Return image size in KB from file on disk."""

    if not os.path.isfile(file_path):
        raise FileNotFoundError(f"File not found: {file_path}")
    
    size_bytes = os.path.getsize(file_path)  # Get size in bytes
    size_kb = size_bytes / 1024  # Convert to KB

    return round(size_kb, 2)
```

## load JPEG image

```python
file_name_original = f'./{rootDir}/Image062.jpg'

with Image(filename=file_name_original) as source:
    display_img_html(source, file_name_original, 'source')
```

Note that this image (from 2009, `Images (2009) - Nokia 6300`) is only `1600x1200` (1.33:1 Aspect Ratio: Commonly known as 4:3 or Standard / Fullscreen) and `2560x1440` (1.78:1 Aspect Ratio: Commonly known as 16:9 or Widescreen) is desired 😐

## save image with new file name (like an idiot)

```python
file_name_step01 = f'./{rootDir}/index-splash-{w1}x{h1}.jpg'
with Image(filename=file_name_original) as img:
    img.save(filename=file_name_step01)

with Image(filename=file_name_step01) as source:
    display_img_html(source, file_name_step01, file_name_step01)
```
